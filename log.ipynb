# 拉普拉斯

```python
import numpy as np
import matplotlib.pyplot as plt

# 设置中文字体
plt.rcParams['font.sans-serif'] = ['Arial Unicode MS', 'SimHei', 'DejaVu Sans']
plt.rcParams['axes.unicode_minus'] = False

# 定义函数 y = e^(-|x|)
def f(x):
    return np.exp(-np.abs(x))

# 生成 x 轴数据
x = np.linspace(-5, 5, 500)
y = f(x)

# 创建图形
plt.figure(figsize=(8, 5))

# 绘制曲线
plt.plot(x, y, color='#d0314e', linewidth=2.5, label=r'$y = e^{-|x|}$')

# 峰值标记（在 x=0 处，y=1）
plt.scatter([0], [1], color='#d0314e', s=80, zorder=5, edgecolors='white', linewidth=1.5)
plt.text(0.1, 1.05, r'$(0, 1)$ 峰值', fontsize=11, color='#d0314e')

# 坐标轴辅助线
plt.axhline(y=0, color='black', linewidth=0.8, alpha=0.3)
plt.axvline(x=0, color='black', linewidth=0.8, alpha=0.3)

# 坐标轴标签与标题
plt.xlabel('x', fontsize=12)
plt.ylabel('y', fontsize=12)
plt.title(r'指数衰减曲线 $y = e^{-|x|}$', fontsize=14, fontweight='400')

# 网格与图例
plt.grid(True, alpha=0.25)
plt.legend(fontsize=11)

# 显示图形
plt.tight_layout()
plt.show()
```

# 高斯核

```python
import numpy as np
import matplotlib.pyplot as plt

# 生成数据
x = np.linspace(-4, 4, 500)
y = np.exp(-(x+1)**2)

# 画图
plt.figure(figsize=(8, 5))
plt.plot(x, y, 'b-', linewidth=2.5)

# 坐标轴辅助线
plt.axhline(y=0, color='black', linewidth=0.8)
plt.axvline(x=0, color='black', linewidth=0.8, linestyle='--')

# 标注关键点
plt.plot(0, 1, 'ro', markersize=8)
plt.text(0.1, 0.92, '(0, 1) 最大值', fontsize=11)

plt.plot(1, np.exp(-1), 'go', markersize=6)
plt.text(1.1, 0.4, f'x=1, y={np.exp(-1):.3f}', fontsize=10)

plt.plot(-1, np.exp(-1), 'go', markersize=6)
plt.text(-2.2, 0.4, f'x=-1, y={np.exp(-1):.3f}', fontsize=10)

plt.xlabel('x', fontsize=13)
plt.ylabel('f(x) = e^(-x²)', fontsize=13)
plt.title('高斯核函数', fontsize=14, fontweight='bold')
plt.grid(True, alpha=0.3)
plt.xlim(-4, 4)
plt.ylim(-0.05, 1.05)

plt.tight_layout()
plt.show()
```
