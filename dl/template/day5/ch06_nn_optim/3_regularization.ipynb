```python
import torch
from torch import nn
```

## 批量标准化

```python
x = torch.randint(0, 10, (3, 2)).float()
```

```python
print(x)
```

```python
bn = nn.BatchNorm1d(2)
```

缩放平移：$$y=γx+β$$
标准化（减均值除方差）是为了保证训练稳定、梯度不消失/爆炸；而最后的缩放平移（$$γ、β$$）是为了防止“过度稳定”导致模型表达能力变差

```python
y = bn(x)
print(y)
```
