```python
import torch
import torch.nn as nn
```

```python
# 创建一个全连接层
linear = nn.Linear(5, 3)
```

## 常数初始化

```python
# 全0
nn.init.zeros_(linear.weight)
print(linear.weight)
```

```python
# 全1
nn.init.ones_(linear.bias)
print(linear.bias)
```

```python
# 常数
nn.init.constant_(linear.weight, 2.5)
print(linear.weight)
```

### 秩初始化

```python
nn.init.eye_(linear.weight)
print(linear.weight)
```

## 随机初始化

### 正态分布

```python
# 标准正态分布
nn.init.normal_(linear.weight)
print(linear.weight)
```

```python
# 一般正态分布
nn.init.normal_(linear.weight, 5, 3)
print(linear.weight)
```

### 均匀分布

```python
nn.init.uniform_(linear.weight)
print(linear.weight)
```

```python
nn.init.uniform_(linear.weight, 10, 20)
print(linear.weight)
```

### Xavier初始化

```python
nn.init.xavier_normal_(linear.weight)
print(linear.weight)
```

```python
nn.init.xavier_uniform_(linear.weight)
print(linear.weight)
```

### He初始化

```python
nn.init.kaiming_normal_(linear.weight)
print(linear.weight)
```

```python
nn.init.kaiming_uniform_(linear.weight)
print(linear.weight)
```

## 模型参数初始化

```python
# 定义模型
model = nn.Sequential(
    nn.Linear(3, 4),
    nn.Tanh(),
    nn.Linear(4, 4),
    nn.Tanh(),
    nn.Linear(4, 2),
    # nn.Softmax(dim=-1),
)
```

```python
# 参数初始化函数
def init_params(layer):
    if isinstance(layer, nn.Linear):
        # nn.init.xavier_normal_(layer.weight)
        nn.init.zeros_(layer.weight)
        nn.init.zeros_(layer.bias)
```

```python
model.apply(init_params)
```

```python
print(model.state_dict())
```

```python
x = torch.randn(4, 3)
y = model(x)
print(y)
```
