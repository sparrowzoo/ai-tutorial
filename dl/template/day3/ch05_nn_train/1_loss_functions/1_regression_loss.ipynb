```python
import torch
import torch.nn as nn
```

```python
# 定义预测评分值（logits）和真实值（target）
logits = torch.tensor([10.1,11.01,12,13,  1400])
target = torch.tensor([10,  11,   12,13, 14])
```

## MAE

```python
# 查看 api example
l1_loss=nn.L1Loss()
loss = l1_loss(logits, target)
print("l1 loss: ", loss)
```

## MSE

```python
l2_loss=nn.MSELoss(reduction='none')
loss = l2_loss(logits, target)
print("l1 loss: ", loss)
```

## Smooth L1

```python
# 注意1/2
smooth_loss=nn.SmoothL1Loss()
loss = smooth_loss(logits, target)
print("smooth_loss loss: ", loss)
```

# 总结
- 第一次尝试时：Smooth L1，它几乎总是表现不错，不会出大错。

- 如果发现模型被离群点严重带偏：换L1。

- 如果你非常确定数据很干净（归一化良好、无异常值）：用L2，收敛更快。
