```python
import torch
from torch import nn
```

```python
# 定义数据：特征和目标值
x = torch.tensor(10.0)
y = torch.tensor(3.0)
print(y.is_leaf)
```

```python
# 随机初始化参数：权重和偏置 并查看requires_grad
w = torch.rand(1, requires_grad=True)
b = torch.rand(1, requires_grad=True)
```

```python
# 模型前向传播，得到预测输出值
z = w * x + b
```

```python
# detach API Returns a new Tensor, detached from the current graph.
#返回一个新张量，从当前计算图中脱离。
#The result will never require gradient.
#结果张量永远不需要梯度。
print(z.detach().requires_grad)
```
