## 其它激活函数（ReLU变体）

```python
# LeakyReLU
#超参数
print(x)
print("ReLU",nn.ReLU()(x))
print("LeakyReLU",nn.LeakyReLU()(x))
```

```python
# PReLU
print("PReLU",nn.PReLU()(x))
# RReLU
print("RReLU",nn.RReLU()(x))
# ELU
print("ELU",nn.ELU()(x))
# SiLU
print("SiLu",nn.SiLU()(x))
print("GELu",nn.GELU()(x))
# Softplus
print("Softplus",nn.Softplus()(x))
```
