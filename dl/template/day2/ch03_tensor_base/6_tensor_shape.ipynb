## 3.7 张量拼接

```python
# 1）torch.cat()：按指定维度拼接（维度不变）
# 核心规则：除拼接维度外，其他维度必须完全相同，拼接后张量维度数不变，仅拼接维度的大小增加。
```

```python
# 2）torch.stack()：堆叠操作（新增维度）
# 本质是先给每个张量在指定位置添加一个大小为 1 的维度（unsqueeze），再按该维度拼接（cat），因此堆叠后张量会新增一个维度。
# 核心要求：所有待堆叠张量的形状必须完全相同。
# 等价于torch.cat([t1.unsqueeze(0), t2.unsqueeze(0)], dim=0)
```
