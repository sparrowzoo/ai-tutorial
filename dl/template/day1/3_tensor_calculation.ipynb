## 3.3.2 元素级运算（张量与张量）

```python
### 严格一致：当两个张量形状完全相同时，直接按位置对应进行运算。
### 广播机制（Broadcasting）：当形状不同时，PyTorch 会自动将较小的张量扩展为与较大张量相同的形状，再进行逐元素运算。
```

## 3.3.3 矩阵相乘

### 3.3.3.1 二维矩阵相乘

```python
tensor1 = torch.tensor(
    [
        [1, 2, 3],
        [4, 5, 6]
    ])
tensor2 = torch.tensor([
    [10,100,1000,10000],
    [20,200,2000,20000],
    [30,300,3000,30000]])
```

```python
print(tensor1 @ tensor2)

 #    def __rmatmul__(self, other: "Tensor") -> "Tensor":
 #        return torch.matmul(other, self)
```

```python
print(tensor1.matmul(tensor2))
```

```python
print(tensor1.mm(tensor2))
```
