```python
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# 设置参数
lambda_val = 0.6
range_val = 2.8
grid_points = 60

# 生成网格
x = np.linspace(-range_val, range_val, grid_points)
y = np.linspace(-range_val, range_val, grid_points)
X, Y = np.meshgrid(x, y)

# L2 正则化: R(w) = 0.5 * lambda * (w1^2 + w2^2)
Z_L2 = 0.5 * lambda_val * (X**2 + Y**2)

# L1 正则化: R(w) = lambda * (|w1| + |w2|)
Z_L1 = lambda_val * (np.abs(X) + np.abs(Y))

# 创建画布
fig = plt.figure(figsize=(14, 6))

# ===== L2 子图 =====
ax1 = fig.add_subplot(121, projection='3d')
ax1.plot_surface(X, Y, Z_L2, cmap='Blues', alpha=0.85, edgecolor='none')
ax1.scatter(0, 0, 0, color='white', s=50, label='原点 (最低点)')
ax1.set_title(r'L2 正则化: $R(w) = \frac{1}{2}\lambda (w_1^2 + w_2^2)$', fontsize=12)
ax1.set_xlabel('$w_1$', fontsize=10)
ax1.set_ylabel('$w_2$', fontsize=10)
ax1.set_zlabel('$R(w)$', fontsize=10)
ax1.view_init(elev=30, azim=45)

# ===== L1 子图 =====
ax2 = fig.add_subplot(122, projection='3d')
ax2.plot_surface(X, Y, Z_L1, cmap='Oranges', alpha=0.85, edgecolor='none')
ax2.scatter(0, 0, 0, color='white', s=50, label='原点 (最低点)')
ax2.set_title(r'L1 正则化: $R(w) = \lambda (|w_1| + |w_2|)$', fontsize=12)
ax2.set_xlabel('$w_1$', fontsize=10)
ax2.set_ylabel('$w_2$', fontsize=10)
ax2.set_zlabel('$R(w)$', fontsize=10)
ax2.view_init(elev=30, azim=45)

plt.tight_layout()
plt.show()
```
