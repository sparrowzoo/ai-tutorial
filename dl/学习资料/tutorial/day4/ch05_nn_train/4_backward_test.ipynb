```python
import torch
from torch import nn
```

```python
# 定义数据：特征和目标值
x = torch.tensor(10.0)
y = torch.tensor(3.0)
```

```python
# 随机初始化参数：权重和偏置
w = torch.rand(1, requires_grad=True)
b = torch.rand(1, requires_grad=True)
print(w)
print(b)
print(w.requires_grad)
print(b.requires_grad)
print(x.requires_grad)
print(y.requires_grad)
```

```python
# 模型前向传播，得到预测输出值
z = w * x + b
print(z)
print(z.requires_grad)
print(z.is_leaf)
```

```python
# 计算损失
loss_fn = torch.nn.MSELoss()
loss = loss_fn(z.squeeze(), y)
print(loss)
```

```python
# 查看参数值和梯度
print(w.data)
print(w.grad)
```

```python
# 反向传播，计算梯度
loss.backward()
```

```python
# 查看参数值和梯度
print(w.data)
print(w.grad)
print(z.grad)
```

```python
# 查看是否叶子节点
print(x.is_leaf)
print(y.is_leaf)
print(w.is_leaf)
print(b.is_leaf)
print(z.is_leaf)
print(loss.is_leaf)
```
