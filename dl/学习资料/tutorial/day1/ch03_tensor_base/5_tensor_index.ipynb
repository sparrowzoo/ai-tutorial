## 3.5.4 布尔索引

````python
# 1. 筛选矩阵（三维张量的元素）
mask = [True, False, False]
print(tensor1[mask])
````

````python
# 筛选满足某种条件的矩阵：中间行、第一个元素大于5的矩阵
print(tensor1)
print(tensor1[:, 2, 0])
mask = tensor1[:, 2, 0] > 5
print(mask)
print(tensor1[mask])
````

````python
# 2. 筛选满足条件的元素（具体数值）：大于5的元素
mask = tensor1 > 5
print(mask)
print(tensor1[mask])
````

````python
# 3. 筛选满足条件的行：首元素大于5
mark_tensor = tensor1[:, :, 3]
mask = mark_tensor > 5
print(tensor1.shape)
print(mask.shape)
print("mark tensor",mark_tensor)
print(mask)
print(tensor1[mask])
````

````python
# 4. 筛选满足条件的列：首元素大于5
mark_tensor = tensor1[:, 0, :]
mask = mark_tensor > 5
print(tensor1.shape)
print(mark_tensor.shape)
print(mask.shape)
print(tensor1.mT.shape)

#print(tensor1.mT[mask].mT)
````

布尔索引从左到右匹配维度，从左边的第 0 维开始依次匹配。

```
数据 shape: [d0, d1, d2, d3]
布尔 mask  : [m0]        → 匹配 m0==d0
布尔 mask  : [m0, m1]    → 匹配 m0==d0, m1=d1
布尔 mask  : [m0, m1, m2]→ 匹配 m0==d0, m1=d1,m2=d2
```
