```python
import torch
```

## 3.3.1 基本运算

```python
tensor1 = torch.randint(1, 10, (2, 3))
print(tensor1)
```

```python
tensor2 = tensor1 + 10
print(tensor2)
print(tensor1)
```

```python
tensor3 = tensor1.add(10)
print(tensor3)
print(tensor1)
```

```python
tensor4 = tensor1.add_(10)
print(tensor4)
print(tensor1)
```

```python
tensor1.add_(10)
print(tensor1)
```

```python
print(-tensor1)
```

```python
print(tensor1 ** 2)
```

```python
print(tensor1.sqrt())
```

## 3.3.2 元素级运算（张量与张量）

```python
tensor1 = torch.tensor([[1, 2, 3], [4, 5, 6]])
tensor2 = torch.tensor([[10, 20, 30], [40, 50, 60]])
print(tensor1)
print(tensor2)
```

```python
print(tensor1 + tensor2)
```

```python
print(tensor1 * tensor2)
```

```python
print(tensor1.mul(tensor2))
```

```python
print(tensor1)
```

```python
print(tensor2 ** tensor1)
```

```python
tensor3 = torch.tensor([10, 20, 50])
print(tensor1)
print(tensor3)
```

```python
print(tensor1 + tensor3)
```

## 3.3.3 矩阵相乘

### 3.3.3.1 二维矩阵相乘

```python
tensor1 = torch.tensor([[1, 2, 3], [4, 5, 6]])
tensor2 = torch.tensor([[1,2,3,4], [5,6,7,8], [6,7,8,9]])
print(tensor1)
print(tensor2)
```

```python
print(tensor1 @ tensor2)
```

```python
print(tensor1.matmul(tensor2))
```

```python
print(tensor1.mm(tensor2))
```

```python
tensor3 = torch.tensor([1,2,3])
print(tensor1 @ tensor3)
```

```python
print(tensor1.matmul(tensor3))
# print(tensor1.mm(tensor3))
```

### 3.3.3.2 高维张量相乘（批量矩阵乘法）

```python
tensor1 = torch.randint(1, 10, (3, 2, 4))
tensor2 = torch.randint(1, 10, (3, 4, 5))
print(tensor1)
print(tensor2)
```

```python
print(tensor1 @ tensor2)
```

```python
print(tensor1.matmul(tensor2))
```

```python
print(tensor1.bmm(tensor2))
```

```python
tensor3 = torch.randn(3, 2, 4, 2)
tensor4 = torch.randn(3, 1, 2, 2)
print(tensor3 @ tensor4)
```
