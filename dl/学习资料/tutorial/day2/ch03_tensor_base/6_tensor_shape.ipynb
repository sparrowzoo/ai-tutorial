```python
import torch
```

```python
tensor1 = torch.randint(1, 10, (2, 3, 6))
print(tensor1)
```

## 3.6.1 维度交换

```python
# .T：翻转所有维度
print(tensor1.T)
print(tensor1.T.shape)
```

```python
# .mT：矩阵转置
print(tensor1.mT)
print(tensor1.mT.shape)
```

```python
# .transpose: 交换任意两个维度
print(tensor1.transpose(0, 1))
```

```python
print(tensor1.transpose(-1, -2))
```

```python
# .permute: 重排列所有维度
print(tensor1.permute(2, 0, 1))
```

## 3.6.3 增删维度

```python
tensor2 = torch.tensor([1, 2, 3, 4, 5])
print(tensor2)
print(tensor2.shape)
print(tensor2.dim())
```

```python
tensor3 = tensor2.reshape(1, -1)
print(tensor3)
print(tensor3.shape)
print(tensor3.dim())
```

```python
# 1）增加维度（unsqueeze()方法）
tensor3 = tensor2.unsqueeze(dim=0)
print(tensor3)
print(tensor3.shape)
```

```python
tensor4 = tensor2.unsqueeze(dim=1)
print(tensor4)
print(tensor4.shape)
```

```python
# 2）删除维度（squeeze()方法）
print(tensor3.squeeze(dim=0))
```

```python
print(tensor3.squeeze(dim=1))
```

```python
print(tensor4.squeeze(dim=1))
```

```python
tensor2.unsqueeze_(dim=0)
tensor2.unsqueeze_(dim=-1)
print(tensor2)
print(tensor2.shape)
```

```python
print(tensor2.squeeze())
```

## 3.7 张量拼接

```python
# 1）torch.cat()：按指定维度拼接（维度不变）
tensor1 = torch.randint(1, 10, (3, 2, 5))
tensor2 = torch.randint(1, 10, (3, 1, 5))
print(tensor1)
print(tensor2)
```

```python
tensor3 = torch.cat([tensor1, tensor2], dim=1)
print(tensor3)
```

```python
tensor1 = torch.randint(1, 10, (4, 2, 5))
tensor2 = torch.randint(1, 10, (2, 2, 5))
print(tensor1)
print(tensor2)
```

```python
print(torch.cat((tensor1, tensor2)))
```

```python
# 2）torch.stack()：堆叠操作（新增维度）
tensor1 = torch.randint(1, 10, (3, 4, 5))
tensor2 = torch.randint(1, 10, (3, 4, 5))
print(tensor1)
print(tensor2)
```

```python
tensor3  = torch.stack([tensor1, tensor2], dim=0)
print(tensor3)
print(tensor3.shape)
```
