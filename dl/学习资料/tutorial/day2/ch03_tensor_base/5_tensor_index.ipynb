```python
import torch
```

```python
tensor1 = torch.randint(1, 10, (3, 5, 6))
print(tensor1)
```

## 3.5.1 简单索引

```python
print(tensor1[1,3,4])
```

```python
print(tensor1[1][3][4])
```

```python
print(tensor1[1])
```

```python
print(tensor1[1,3])
```

```python
print(tensor1[0, -1])
```

```python
print(tensor1[:, 0])
```

## 3.5.2 范围索引

```python
print(tensor1[1:])
```

```python
print(tensor1[0:-1, 1:4:2, 0:4:2])
```

```python
print(tensor1[-1])
```

```python
print(tensor1[-1:])
```

## 3.5.3 列表索引

```python
list1 = [0, 1, 1]
list2 = [2, 1, 3]
list3 = [1, 5, 3]
print(tensor1[list1, list2, list3])
```

```python
print(tensor1[[[0], [1]], [1, 2]])
```

## 3.5.4 布尔索引

```python
# 1. 筛选矩阵（三维张量的元素）
mask = [True, False, True]
print(tensor1[mask])
```

```python
# 筛选满足某种条件的矩阵：中间行、第一个元素大于5的矩阵
mask = tensor1[:, 2, 0] > 5
print(mask)
print(tensor1[mask])
```

```python
# 2. 筛选满足条件的元素（具体数值）：大于5的元素
mask = tensor1 > 5
print(mask)
```

```python
print(tensor1[mask])
```

```python
# 3. 筛选满足条件的行：首元素大于5
mask = tensor1[:, :, 0] > 5
print(mask)
```

```python
print(tensor1[mask])
```

```python
# 4. 筛选满足条件的列：首元素大于5
mask = tensor1[:, 0, :] > 5
print(mask)
```

```python
print(tensor1.mT[mask].mT)
```
