```python
import torch
import torch.nn as nn
```

## 构造方法

```python
linear = nn.Linear(in_features=2, out_features=3)
```

```python
print(linear)
```

```python
# 查看线性层的参数
print("权重：", linear.weight)
print("偏置：", linear.bias)
```

## 前向传播 

```python
# 定义输入数据（2个特征）
x = torch.tensor([1, 2]).float()
```

```python
# 调用方法，得到输出y
y = linear.forward(x)
print(y)
```

```python
# 定义多条数据（5条）
X = torch.randn(5, 2)
```

```python
Y = linear.forward(X)
print(Y)
```

```python
# 隐式调用
y = linear(x)
print(y)
Y = linear(X)
print(Y)
```
