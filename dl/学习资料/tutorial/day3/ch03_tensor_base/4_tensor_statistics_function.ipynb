```python
import torch
```

```python
tensor1 = torch.randint(1, 10, (3, 2, 4))
print(tensor1)
```

```python
# sum
print(tensor1.sum())
```

```python
print(tensor1.sum(dim=1))
```

```python
print(tensor1.sum(dim=-1))
```

```python
print(tensor1.sum(dim=0))
```

```python
tensor1 = tensor1.float()
```

```python
# mean
print(tensor1.mean(dim=0))
```

```python
# var
print(tensor1.var(dim=1))
```

```python
# std
print(tensor1.std(dim=2))
```

```python
# max/min
print(tensor1.max(dim=2))
```

```python
# argmax
print(tensor1.argmax(dim=2))
```

```python
# 去重
print(tensor1.unique(return_counts=True))
```

```python
# 排序
print(tensor1.sort())
```
