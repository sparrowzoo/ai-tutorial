```python
# 7. float64
tensor1 = torch.tensor([1, 2, 3], dtype=torch.float64)
print(tensor1)
print(tensor1.dtype)
tensor2 = torch.DoubleTensor([1, 2, 3])
print(tensor2)
print(tensor2.dtype)
```

## 3.1.5 根据数据区间创建张量

```python
# 1）固定步长生成
tensor1 = torch.arange(10)
print(tensor1)
```

```python
tensor2 = torch.arange(2, 20, 3)
print(tensor2)
```

```python
print(tensor2.dtype)
```

```python
# 2）固定数量线性生成
tensor1 = torch.linspace(10, 30, 5)
print(tensor1)
```

```python
print(tensor1.dtype)
```

```python
# 3）对数间隔生成
tensor1 = torch.logspace(1, 3, 3)
print(tensor1)
```

```python
tensor2 = torch.logspace(1, 3, 3, base=2)
print(tensor2)
```

## 3.1.6 根据填充创建张量

```python
# 1）全 1 张量
tensor1 = torch.ones(2, 3, dtype=torch.int32)
print(tensor1)
```

```python
tensor2 = torch.ByteTensor(3, 4)
print(tensor2)
```

```python
tensor3 = torch.ones_like(tensor2)
print(tensor3)
```

```python
# 2) 全0
tensor1 = torch.zeros(2, 3, dtype=torch.int16)
print(tensor1)
```

```python
tensor2 = torch.zeros_like(tensor3)
print(tensor2)
```

```python
# 3）未初始化张量
tensor1 = torch.empty(3, 5)
print(tensor1)
```

```python
# 4）全常量张量
tensor1 = torch.full((3, 2), 10.5)
print(tensor1)
```

```python
# 5）单位矩阵
tensor1 = torch.eye(4)
print(tensor1)
```

```python
tensor2 = torch.eye(4, 3)
print(tensor2)
```

## 3.1.7 随机创建张量

```python
# 1）均匀分布随机张量
# 浮点数
tensor1 = torch.rand(2, 3, dtype=torch.float16)
print(tensor1)
```

```python
tensor2 = torch.rand_like(tensor1)
print(tensor2)
```

```python
# 整数
tensor3 = torch.randint(10, 100, [5, 7])
print(tensor3)
```

```python
# 2）正态分布随机张量
tensor1 = torch.randn(3, 2, 5)
print(tensor1)
```

```python
tensor2 = torch.normal(5, 2, tensor1.shape)
print(tensor2)
```

```python
# 3）随机打乱与随机种子
tensor1 = torch.randperm(10)
print(tensor1)
```

```python
seed = torch.random.initial_seed()
print(seed)
```
