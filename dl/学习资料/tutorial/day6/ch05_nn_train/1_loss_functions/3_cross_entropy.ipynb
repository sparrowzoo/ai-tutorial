```python
import torch
from torch import nn
```

```python
# 定义输出预测值
logits = torch.randn(10, 6)
print(logits)
```

```python
# 定义损失函数
loss_fn = nn.CrossEntropyLoss()
```

### 真实值为标签编码

```python
target = torch.randint(0, 6, (10,))
print(target)
```

```python
loss = loss_fn(logits, target)
print(loss)
```

### 真实值为独热编码

```python
target_onehot = torch.zeros(10, 6)
target_onehot[torch.arange(10), target] = 1
print(target_onehot)
```

```python
loss = loss_fn(logits, target_onehot)
print(loss)
```
