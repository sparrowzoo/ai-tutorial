```python
import torch
from torch import nn
```

## 批量标准化

```python
x = torch.randint(0, 10, (3, 2)).float()
```

```python
print(x)
```

```python
bn = nn.BatchNorm1d(2)
```

```python
y = bn(x)
print(y)
```

```python
optimizer = torch.optim.SGD(model.parameters(), lr=0.01, weight_decay=0.1)
```

```python
optimizer2 = torch.optim.Adam(model.parameters(), lr=0.01, weight_decay=0.1)
optimizer3 = torch.optim.AdamW(model.parameters(), lr=0.01)
```
