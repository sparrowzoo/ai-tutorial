```python
import torch
import numpy as np
```

## 3.2.1 元素类型转换

```python
tensor1 = torch.tensor([1, 2, 3])
print(tensor1)
print(tensor1.dtype)
```

```python
# 方式1：type() 方法
tensor2 = tensor1.type(torch.float16)
print(tensor2, tensor2.dtype)
```

```python
# 方式2：to() 方法
tensor3 = tensor1.to(torch.float16)
print(tensor3, tensor3.dtype)
```

```python
# 方式3：直接调用类型方法
tensor4 = tensor1.half()
print(tensor4, tensor4.dtype)
```

```python
tensor5 = tensor1.float()
print(tensor5, tensor5.dtype)
```

## 3.2.2 Tensor与numpy的ndarray转换

```python
np.set_printoptions(precision=6)
torch.set_printoptions(precision=6)
```

### 3.2.2.1 tensor => ndarray

```python
tensor1 = torch.randn(3, 2)
print(tensor1)
```

```python
arr1 = tensor1.numpy()
print(arr1)
```

```python
print(tensor1[0,0])
```

```python
tensor1[0,0] = 10
```

```python
print(tensor1)
```

```python
print(arr1)
```

```python
# 拷贝数据避免内存共享
arr2 = tensor1.numpy().copy()
print(arr2)
```

```python
tensor1[0,1] = 25
```

```python
print(tensor1)
print(arr1)
```

```python
print(arr2)
```

### 3.2.2.2 NumPy ndarray => Tensor

```python
arr1 = np.random.randn(3, 2)
print(arr1)
```

```python
print(arr1.dtype)
```

```python
tensor1 = torch.from_numpy(arr1)
print(tensor1)
```

```python
arr1[0,0] = 13
```

```python
print(arr1)
print(tensor1)
```

```python
# 拷贝避免数据内存共享
tensor2 = torch.from_numpy(arr1.copy())
print(tensor2)
```

```python
tensor3 = torch.from_numpy(arr1).clone()
print(tensor3)
```

```python
tensor4 = torch.tensor(arr1)
print(tensor4)
```

```python
arr1[0, 0] = 32
print(arr1)
print(tensor1)
```

```python
print(tensor2)
print(tensor3)
print(tensor4)
```

## 3.2.3 张量（Tensor）=> 标量（Scalar）

```python
tensor1 = torch.tensor(10)
print(tensor1)
```

```python
print(tensor1.dim())
print(tensor1.shape)
```

```python
scalar = tensor1.item()
print(scalar)
print(type(scalar))
```

```python
tensor2 = torch.tensor([10])
print(tensor2)
print(tensor2.dim())
print(tensor2.shape)
```

```python
scalar = tensor2.item()
print(scalar)
print(type(scalar))
```

```python
tensor3 = torch.tensor([[10]])
print(tensor3)
print(tensor3.dim())
print(tensor3.shape)
```

```python
scalar = tensor2.item()
print(scalar)
print(type(scalar))
```
