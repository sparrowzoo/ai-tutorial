# 张量的聚合操作

```python
import torch
t1 = torch.randint(1, 10, (2, 3, 4))
print("原始张量 t1：")
print(t1)
print(f"张量形状：{t1.shape}")
```

```python
t1.sum()
```

```python
t1.sum(dim=0)
```

```python
t1.sum(dim=1)
```

```python
t1.sum(dim=2)
```

```python
#求均值
t1.float().mean(dim=1)
```

```python
# 最大最小API
t1.max(dim=1)
```

```python
print("最大位置信息",t1.argmax(dim=1))
print("最小值位置信息",t1.argmin(dim=1))
```

```python
print(t1)
var1=t1.float().var(dim=1,unbiased=False)
print(t1[0][:,0])
dim1=t1[0][:,0]
dim1_mean=dim1.float().mean()
print("dim1 mean",dim1_mean)
print(((dim1-dim1_mean)**2).sum()/3)
print("t1' var",var1)
```

```python
t1=torch.tensor([[1,2,3],[4,5,6]])
print("t1.shape",t1.dim())
print(t1.sum(dim=1,keepdim=True).dim())

print("t1",t1)
print(t1.sum(dim=1,keepdim=True))
```

```python
t1 = torch.randint(1, 10, (2, 3, 4))
print("t1",t1)
t_sum0=t1.sum(dim=0)
print("t_sum0",t_sum0)
```

```python
t_sum1=t1.sum(dim=1)
print("t_sum1",t_sum1)
```

```python
t_sum2=t1.sum(dim=2,keepdim=False)
print("t_sum2",t_sum2.shape)
```

```python
#（3）最值运算：max() / min()
```

```python
print(t1)
t1_max,t1_max_indices=t1.max(dim=1)
print("t1_max",t1_max)
print("t1_max indices",t1_max_indices)

print("t1 argmax",t1.argmax(dim=1))
print("t1 argmin",t1.argmin(dim=1))
```

```python
# （4）方差与标准差：var() / std()
t1_float=t1.float()
print("t1_float",t1_float)
var1=t1_float.var(dim=1,unbiased=False)
print("var1",var1)
```

```python
l11=t1_float[0][:,0]
print("第0个矩阵的第0行",l11)
l11_mean=l11.mean()
print("l11_mean",l11_mean)
```

```python
print(((l11-l11_mean)**2).sum()/3)
```

```python
#（5）去重运算：unique()
t1=torch.randint(1,10,(2, 3, 4))
print("t1 unique",t1.unique(sorted=False,return_counts=False))
```

```python
#（6）排序运算：sort()
t1.sort(dim=2, descending=True)
```

```python
t1.float().var(dim=2,unbiased=False,keepdim=True).shape
```

```python
print(t1.float().max(dim=2,keepdim=True).values.shape)
```
