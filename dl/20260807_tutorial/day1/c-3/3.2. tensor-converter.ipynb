# 3.2张量（Tensor）转换

## 3.2.1元素类型转换

```python
import torch
tensor1=torch.tensor([[1,2,3],[4,5,6]])
print(tensor1)
print("tensor1 type",tensor1.dtype)
```

```python
# type
tensor2=tensor1.type(dtype=torch.int8)
print("tensor2 type",tensor2.dtype)
tensor3=tensor1.type(dtype=torch.float)
print("tensor3 type",tensor3.dtype)
# to
tensor22=tensor1.to(dtype=torch.int8,device='cpu')
print("tensor22 type",tensor22.dtype)
tensor32=tensor1.to(dtype=torch.float)
print("tensor32 type",tensor32.dtype)
# .float
tensor41=tensor1.float()
print("tensor41 type",tensor41.dtype)
```

## 3.2.2Tensor与numpy的ndarray转换

### 3.2.2.1tensor => ndarray

```python
import numpy as np
# tensor -- ndarray
tensor1=torch.tensor([[1,2,3],[4,5,6]])
```

```python
array=tensor1.numpy()
print("array type",type(array))
print("numpty array",array)
print("tensor1 array",tensor1)
```

```python
array[0][0]=100
```

```python
print("numpty array",array)
print("tensor1 array",tensor1)
```

```python
tensor1[0][1]=1000
```

```python
print("numpty array",array)
print("tensor1 array",tensor1)
```

```python
tensor2=tensor1.clone()
array2=tensor2.numpy()
tensor1[0][2]=99999
print("numpty array2",array2)
print("tensor1 array",tensor1)
print("tensor2 array",tensor2)
```

```python
# ndarray -- tensor
array=np.array([1,2,3])
tensor1=torch.from_numpy(array)
array[1]=9999
print(" array",array)
print("tensor2 array",tensor1)
```

```python
array2=array.copy()
array[2]=888
print(" array",array)
print("array2",array2)
```

## 3.2.3张量（Tensor）=> 标量（Scalar）

```python
tensor1=torch.tensor([[[8]]])
print("张量转标量",tensor1.item())
```

```python
tensor2=tensor1.item()
tensor1[0][0]=999
```

```python
print("tensor1 item",tensor1)
print("tensor2 item",tensor2)
```
