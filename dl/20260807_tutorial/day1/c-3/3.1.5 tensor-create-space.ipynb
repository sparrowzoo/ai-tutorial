# 基于数据区间创建张量

## 2）固定步长线性生成：torch.linspace

```python
import torch
#指定步长创建张量
#start end step
print(torch.arange(10))# end=10 start=? default 0,step default 1
print(torch.arange(1,10,2))

print(torch.arange(1,10.1,1.1)) #支持小数步长
```

## 2）固定数量线性生成：torch.linspace

```python
# 参数没有默认值
#step=(10-0)/(10-1)-1.11
print("0~10",torch.linspace(0,10,10)) #start=0,end=10 steps=?
#step=(9-0)/(10-1)=1
print("0~9",torch.linspace(0,9,10))
```

3）对数间隔生成：torch.logspace

```python
## 对数间隔分布
#以e 为底
torch.logspace(1,3,3,base=torch.e)
#2底
torch.logspace(1,3,3,base=2)

torch.linspace(1)
```
