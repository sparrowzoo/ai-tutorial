## 简单索引

```python
import torch
from numpy.ma.core import masked

t1=torch.randint(1,100,(2,5,4))
print(t1)
```

```python
# 第0个矩阵 第0行 第0列
t1[0][0][0]
```

```python
t1[0,0,0]
```

```python
t1[0,0,0:400]
```

```python
t1[0,0,0:-1]
```

```python
# 1. 取第1个矩阵（dim=0=1），从第1行（dim=1=1）到末尾的所有行、所有列
# 等价于 t[1, 1:, :]，输出形状 (4,4)
t1[1,1:,:]
```

```python
# 2. 复杂切片：从倒数第1个矩阵（dim=0=-1:）开始，
# 行维度：从第1行（包含）到第4行（不包含），即行索引1、2、3
# 列维度：从第0列到第2列（不包含），步长2，即列索引0

print("t1 切片",t1[-1,1:4,0:2:2])
print(t1)
```

## 列表索引

```python
# 1. 一维列表索引：按位置一一对应提取元素
# list1 对应 dim=0（矩阵维度），list2 对应 dim=1（行维度）
# 提取位置：(0,1)、(1,3)、(1,2)，最终取对应位置的整行（dim=2 全取）
list1 = [0, 1, 1]
list2 = [1, 3, 2]
print("=== 一维列表索引 t[list1, list2] ===")
print(t1[list1, list2])
print(t1)
```

```python
        [[[40, 83, 47, 14],
         [53, 89, 17, 92],
         [70, 99, 65, 62],
         [85, 28, 99, 57],
         [14,  5, 89, 52]],

        [[ 9, 10, 33, 19],
         [39, 56, 85, 14],
         [22, 99, 80,  1],
         [ 7, 67, 76, 57],
         [35, 86, 93, 42]]]
```

```python
# 2. 二维列表索引：广播匹配，批量提取多维度元素
# list1 为 [[0], [1]]（对应2个矩阵），list2 为 [1,4,3]（对应3行）
# 广播后提取：第0个矩阵的第1、4、3行，第1个矩阵的第1、4、3行
list1 = [[0], [1]]
print(torch.tensor(list1).shape)
list2 = [[1, 4, 3]]
print(torch.tensor(list2).shape)

list1 = [[0,0,0], [1,1,1]]
print(torch.tensor(list1).shape)
list2 = [[1, 4, 3],[1, 4, 3]]
print(torch.tensor(list2).shape)


print("\n=== 二维列表索引 t[list1, list2] ===")
print(t1[list1, list2])
print(f"输出形状：{t1[list1, list2].shape}") # 输出形状 (2,3,4)
```

```python
# 1. 取出所有大于5的值
mark1=t1>5
print(t1)
print(mark1)
print("mask shape",mark1.shape)
print("t1 shape",t1.shape)
print("t1[mark]",t1[mark1])
print("t1 mark shape",t1[mark1].shape)
```

```python
# 2. 取出满足条件的行: 每行的首元素大于5
mask2=t1[:,:,0]>50
print("mask:",mask2)
print("mask shape",mask2.shape)
print("t1 shape",t1.shape)
print("t1 满足条件 的行",t1[mask2])

print("t1",t1)
```

```python
# 3. 取出满足条件的列: 每列的首元素大于5
mark3=t1[:,0,:]>5
print(mark3.shape)
```

```python
print(t1.shape)
t1.mT[mark3].mT
```

```python
# 4. 选取符合条件的矩阵：(1, 2) 元素小于1
mask4=t1[:,1,2]>84
print(mask4.shape)
print(t1.shape)
```
