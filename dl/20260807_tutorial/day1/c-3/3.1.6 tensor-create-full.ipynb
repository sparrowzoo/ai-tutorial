# 3.1.6根据填充创建张量

## 1）全 1 张量：torch.ones & torch.ones_like

```python
import torch
##生成【2，3】 的全1的张量
torch.ones(2,3)
tensor=torch.tensor([[1,2,3,4],[4,5,6,7]])
print("原始张量",tensor)
```

```python
tensor1=torch.ones_like(tensor)
print("创建like 张量",tensor1)
```

```python
print("全0张量",torch.zeros(2,3))
tensor0=torch.zeros_like(tensor)
print("创建like 0 张量",tensor0)
print("创建like 0 张量形状",tensor0.shape)
```

```python
tensor2=torch.tensor([1.0])
print("like tensor2",torch.zeros_like(tensor2))

print("tensor2 type",tensor2.dtype)
print("like tensor2 type",torch.zeros_like(tensor2).dtype)
```

## 2）未初始化张量：torch.empty

```python
empty_tensor=torch.empty(3,5)
print("空张量",empty_tensor)
print("空张量的类型",empty_tensor.dtype)

empty_tensor2=torch.Tensor(3,5)
print("空张量2",empty_tensor2)
print("空张量2的类型",empty_tensor2.dtype)

empty_tensor[0]=torch.ones(1)
print("empty_tensor",empty_tensor)
```

## 3）全常量张量：torch.full

```python
torch.full([1,10],1.0).dtype
```

```python
torch.full((1,10),10).dtype
```

## 4）单位矩阵：torch.eye

```python
eye1=torch.eye(3)
print("3x3 的方阵",eye1)
print("类型",eye1.dtype)
```

```python
eye2=torch.eye(5,4)
print("5x4的单位矩阵",eye2)
print("类型，shape",eye2.dtype,eye2.shape)
```

```python
eye3=torch.eye(5,8)
print("5x8的单位矩阵",eye3)
print("类型，shape",eye3.dtype,eye3.shape)
```

```python
tensor3=torch.tensor([1.0])
print(torch.ones_like(tensor3).dtype)
```
