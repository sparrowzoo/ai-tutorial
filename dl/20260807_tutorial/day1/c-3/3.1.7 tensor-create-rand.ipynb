# 随机创建张量

## 1）均匀分布随机张量

```python
import torch
## 0~1 浮点数随机
torch.rand(1,2,dtype=torch.float16)
```

```python
torch.randint(0,10,(2,3,4))
```

```python
torch.randint(10,11,(1,))
```

## 2）正态分布随机张量

```python
torch.randn(2,3)
```

```python
torch.normal(2,3,[2,3])
```

## 3）随机打乱与随机种子

```python
torch.randperm(1)
```

### 随机种子

```python
seed=torch.initial_seed()
print("随机种子",seed)
```

```python
torch.manual_seed(42)
```

```python
seed=torch.initial_seed()
print("随机种子",seed)
```

```python
torch.manual_seed(43)
print("第一轮生成",torch.rand(1,2))
print("第一轮生成",torch.rand((2,3)))
torch.manual_seed(43)
print("第二轮生成",torch.rand(1,2))
print("第二轮生成",torch.rand((2,3)))

```

```python
torch.manual_seed(44)
print("第一轮生成",torch.rand(1,2))
print("第一轮生成",torch.rand((2,3)))
```
