# 加载数据

```python
import pandas as pd
import torch
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import MinMaxScaler
```

```python
samples=pd.read_csv('./train.csv')
print(type(samples))
samples.head()
```

```python
## 拆分标签样本
labels=samples.iloc[:,0]
features=samples.iloc[:,1:]
print("labels:",labels.shape)
print("features:",features.shape)
```

# 拆分数据集

```python
features_trains,features_test,labels_trains,labels_tests=train_test_split(features,labels,test_size=0.2,random_state=42)

print("训练集特征shape",features_trains.shape)
print("测试集特征shape",features_test.shape)

print("训练集标签shape",labels_trains.shape)
print("测试集shape",labels_tests.shape)
```

```python
## 验证数据集类型
print(type(features_trains))
print(type(labels_trains))
```

# 归一化

```python
min_max_scalar=MinMaxScaler()
```

```python
print(min_max_scalar.data_min_)
```

```python
min_max_scalar.fit(features_trains)
print(min_max_scalar.data_min_)
print(min_max_scalar.data_max_)

```

```python
print(type(features_trains))
features_trains=min_max_scalar.fit_transform(features_trains)
print(type(features_trains))

features_test=min_max_scalar.transform(features_test)
```

# 转换tensor

```python
# 转换训练集
features_trains_tensor=torch.tensor(features_trains)
features_trains_tensor=features_trains_tensor.float()
labels_trains_tensor=torch.tensor(labels_trains.values)
# 转换测试集
features_tests_tensor=torch.tensor(features_test).float()
labels_tests_tensor=torch.tensor(labels_tests.values)
```

# 模型加载

## 模型结构定义

```python
import torch.nn as nn
digital_model=nn.Sequential(
    nn.Linear(28*28,50),
    nn.ReLU(),
    nn.Linear(50,100),
    nn.ReLU(),
    nn.Linear(100,10)
)
```

## 加载已存在的模型

```python
state_dict=torch.load("./nn_example.pt",map_location="cpu")
print("state_dict",state_dict)
```

```python
## 加载模型
digital_model.load_state_dict(state_dict)
print(digital_model[0].weight)
print(features_tests_tensor.dtype)
```

# 评估

```python
print("实际标签\n",labels_tests_tensor[0:10])
```

```python
print(features_tests_tensor.dtype)
pred_labels=digital_model(features_tests_tensor)
print("预测标签\n",pred_labels.argmax(dim=1))
```

```python
print(type(labels_tests_tensor[0:10]))

print(pred_labels.argmax(dim=-1))
```

```python
acc_count=labels_tests_tensor.eq(pred_labels.argmax(dim=-1)).sum().item()
```

```python
acc=acc_count/len(pred_labels)
```
