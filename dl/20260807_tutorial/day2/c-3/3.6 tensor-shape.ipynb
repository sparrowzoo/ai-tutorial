## 3.6.2调整形状

```python
import torch
t1=torch.randint(1,100,(2,5,4))
print("t1",t1)
print("t1. shape",t1.shape)
print("t1. element. num",t1.numel())
```

```python
t2=t1.reshape(5,8)
print("t2. shape",t2.shape)
```

```python
t3=t1.reshape(8,-1)
print("t3. shape",t3.shape)
```

```python
t4=t1.view(8,-1)
print("t4. shape",t4.shape)
```

```python
print("t1. continue?",t1.is_contiguous())
```

```python
t5=t1.mT
print("t5 is contiguous",t5.is_contiguous())
```

```python
t5.view(8,5)
```

```python
print("t1 stride",t1.stride())
print("t1 shape",t1.shape)
print("t1 contiguous",t1.is_contiguous())
print("t5 stride",t5.stride())
print("t5 ",t5)

print("t5 contiguous",t5.is_contiguous())
```

```python
t5=t5.contiguous()
t5.view(8,5)
print("t5 is contiguous",t5.is_contiguous())
print("t5 shape",t5.shape)
print("t5 stride",t5.stride())
```

## 3.6.3增删维度

```python
#增加维度
print("t1 shape",t1.shape)
t2=t1.unsqueeze(dim=1)
print("t2. shape",t2.shape)
print("t2",t2)
```

```python
t2=t2.unsqueeze(dim=0)
print("t2",t2.shape)
```

```python
t2=t2.squeeze(dim=0)
print("t2",t2.shape)
```

```python
t2=t2.squeeze()
print("t2",t2.shape)
```
