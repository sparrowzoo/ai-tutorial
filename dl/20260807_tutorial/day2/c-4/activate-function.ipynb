# 激活函数

## 常见的激活函数

### sigmoid

```python
import torch
import torch.nn as nn
x = torch.tensor([0.7, -1.2, 2.5, 0, -3.1])
```

```python
y1=torch.sigmoid(x)
print("torch.sigmoid(x) =",y1)
```

```python
y2=x.sigmoid()
print("x.sigmoid() =",y2)
sigmoid=nn.Sigmoid()
y3=sigmoid.forward(x)
print("Sigmoid() =",y3)
y4=sigmoid(x)
print("sigmoid(x) =",y4)
```

```python
### tanh
y1=torch.tanh(x)
print("torch.tanh(x) =",y1)
```

```python
y2=x.tanh()
print("x.tanh() =",y2)
```

```python
tanh=nn.Tanh()
y3=tanh(x)
y3=tanh.forward(x)
print("tanh(x) =",y3)
```

```python
### relu
y1=torch.relu(x)
print("torch.relu(x) =",y1)
```

```python
y2=x.relu()
relu=nn.ReLU()
y3=relu(x)
y4=relu.forward(x)
print("relu(x) =",y3)
```

```python
## softmax
print("x",x)
y1=torch.softmax(x,dim=0)
print("softmax(x) =",y1)
print(y1.sum())
```

```python
x=torch.tensor([[ 0.7000, -1.2000,  2.5000],[ 1.7000, 2.2000,  3.5000]])
print(x)
```

```python
y2=torch.softmax(x,dim=-1)
print("softmax(x) =",y2)
print(y2.sum(dim=-1))

print("max arg",y2.argmax(dim=-1))
```

```python
nn.Softmax()
x.softmax()
```

## 其他激活函数的使用

```python
x = torch.tensor([0.7, -1.2, 2.5, 0, -3.1])
print(x*0.25)
print("LeakRelu",nn.LeakyReLU()(x))
print("nn.PReLU()",nn.PReLU()(x))
print("nn.RReLU()",nn.RReLU()(x))
print("nn.ELU()",nn.ELU()(x))

print("nn.SiLU()",nn.SiLU()(x))
print("nn.GELU()",nn.GELU()(x))
print("nn.Softplus()",nn.Softplus()(x))
```
