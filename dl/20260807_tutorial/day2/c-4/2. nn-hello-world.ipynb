# 手写简单的神经网络

```python
# 定义网络结构
import torch
import torch.nn as nn
```

```python
# 定义模型
class HelloWorldModel(nn.Module):
    def __init__(self):
        ## 帮我们初始化模型
        super().__init__()
        # 定义第一个隐藏层 3个输出特征 4个输出神经元
        self.linear1 = nn.Linear(in_features=3,out_features=4,bias=True) #W.shape=4*3  12+4=16
        # 定义第二个隐藏层 4个输入 4个输出
        self.linear2 = nn.Linear(in_features=4,out_features=4,bias=True) #4*4+4=20
        # 定义输出层 4个输出 2个输出
        self.out=nn.Linear(in_features=4,out_features=2,bias=True) #4*2+2=10

    # 定义前向传播 定义层的流转
    def forward(self,x):
        #第一层的原始输出
        y1=self.linear1(x)
        #第一个隐藏的激活函数tanh
        y1_act=torch.tanh(y1)
        # 第二层的输出
        y2=self.linear2(y1_act)
        # 第二个隐藏的激活函数relu
        y2_act=torch.relu(y2)
        #输出层
        y_out=self.out(y2_act)
        y=torch.softmax(y_out,dim=-1)
        return y
```

```python
# 前向传播 直接推理
x=torch.randint(1,100,(10,3))
print("样本特征",x)
print("x的类型",x.dtype)
```

```python
# 预测
hello_world=HelloWorldModel()
x=x.float()
x=x.to(dtype=torch.float)
y=hello_world.forward(x)
```

```python
y.sum(dim=-1)
print("输出结果",y)
```

```python
# 4.4.2查看模型参数和模型结构
print("第一层的参数",hello_world.linear1.weight)
print("第一层的参数形状",hello_world.linear1.weight.shape)
print("第一层的偏置",hello_world.linear1.bias)
print("第一层的偏置的形状",hello_world.linear1.bias.shape)
```

```python
## 第二层的参数
print("第二层的参数",hello_world.linear2.weight)
print("第二层的参数形状",hello_world.linear2.weight.shape)
print("第二层的偏置",hello_world.linear2.bias)
print("第二层的偏置的形状",hello_world.linear2.bias.shape)
```

```python
# 用parameters 方法输出所有模型参数
paras=hello_world.parameters()
for param in paras:
    print(param)
```

```python
# 用named parameters 方法输出所有模型参数
named_parameters=hello_world.named_parameters()
for name,param in named_parameters:
    print(name,param)
```

```python
state_dict=hello_world.state_dict()
for name,param in state_dict.items():
    print(name,param)
```
