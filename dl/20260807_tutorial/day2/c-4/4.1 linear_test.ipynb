```python
import torch
import torch.nn as nn
linear=nn.Linear(in_features=3,out_features=4,bias=False)
print(linear.weight.shape)
print(linear.bias)
```

```python
# 前向传播
x=torch.tensor([[1,2,3],[4,5,6]])
x=x.float()
y=linear.forward(x)
print("前向传播结果",y)
print("结果的形状",y.shape)
```

```python
## __call__
y1=linear(x)
print(y1)
```
