```python
import torch
import torch.nn as nn
```

```python
class MyModule(nn.Module):
    def __init__(self):
        super().__init__()
        self.fc1 = nn.Linear(784, 256).to("mps")
        self.fc2 = nn.Linear(256, 128).to("mps")
        self.out = nn.Linear(128, 64).to("mps")

    def forward(self,x):
        x = self.fc1(x)
        x = torch.relu(x)
        x = self.fc2(x)
        x = torch.relu(x)
        x = self.out(x)
        x=torch.sigmoid(x)
        return x
```

```python
test=MyModule()
x=torch.randn(1,784).to("mps")
t1=test(x)
y=test.forward(x)
print(y.device)
```

```python
from torchsummary import summary
from torchinfo import summary
test=test.to(device="mps")
summary(test, input_size=(784,), batch_size=1000,device="mps")
print(test.fc1.weight.device)
```
