## 3.1.1 基于内容创建张量

```python
# 1. 基于标量
tensor1 = torch.tensor(10)
print(tensor1)
```

```python
print(type(tensor1))
print(tensor1.dtype)
tensor1=torch.tensor(1.0)
print(tensor1.dtype)
```

```python
# 查看Tensor的维度和形状
print(tensor1.dim())
```

### 总结 标量是0维！

```python

print(tensor1.size())
print(tensor1.shape)
```

```python
# 2. 基于向量
tensor2 = torch.tensor([1, 2, 3])
print(tensor2)
```

```python
print(tensor2.dim())
print(tensor2.shape)
```

```python
print(tensor2.t())
```

### 总结 向量是1维！不管是行向量还是列向量，都是1维的。不区分行向量 和列向量

```python
# 3. 基于矩阵(二维)
tensor3 = torch.tensor([[1, 2, 3], [4, 5, 6]])
print(tensor3)
```

```python
tensor3 = torch.tensor([[1, 2, 3], [4, 5, 6, 7]])
print(tensor3)
```

```python
print(tensor3.dim())
print(tensor3.shape)
```

### 二维张量是深度学习中批量数据的标准存储形式，就是我们常说的矩阵，
### 例如批量输入的图像特征、数据库的二维表，是这些数据的典型表示形式。

```python
# 4. 基于高维数组
tensor4 = torch.tensor([[[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]], [[1, 2, 3], [4, 5, 6]]])
print(tensor4)
```

```python
print(tensor4.dim())
print(tensor4.shape)
```

### 三维及以上 在深度学习中，我们经常需要处理三维、四维甚至更高维度的数据（如彩色图像、视频序列、批量图像等 在pytorch 中所有的维度都是张量，统一了数据的表示形式，方便我们进行各种操作。为什么？ 怎么理解shape?

```python
# 5. 基于numpy数组
arr = np.array([[1, 2, 3, 4], [4, 5, 6, 7]])
print(arr)
```

```python
tensor5 = torch.tensor(arr)
print(tensor5)
```

```python
print(tensor5.dim())
print(tensor5.shape)

arr[0][0] = 10000
print(tensor5)
```

###  from_numpy 改变原数组

```python
tensor6 = torch.from_numpy(arr)
print(tensor6)
#arr[0][0] = 10000
print(tensor6)
```

## 3.1.5 根据数据区间创建张量

均匀区间、线性间隔、对数间隔的数值序列生成需求，提供了 arange、linspace、logspace 等核心 API

Python 的设计哲学是 "只有一种明显的方式去做一件事"。

```python
# 1）固定步长生成
tensor1 = torch.arange(10) #arange(start=0, 10, step=1
#[0,10)
# for i=0;i<10;i++
# arange(start=0, end, step=1)
#  Example::
#         >>> torch.arange(5)
#         tensor([ 0,  1,  2,  3,  4])
#         >>> torch.arange(1, 4)
#         tensor([ 1,  2,  3])
#         >>> torch.arange(1, 2.5, 0.5)
#         tensor([ 1.0000,  1.5000,  2.0000])
print(tensor1)
```

```python
tensor2 = torch.arange(2, 20, 3)
# for i =2; i < 20; i+=3:
print(tensor2)
```

```python
print(tensor2.dtype)
```

```python
# 2）固定数量线性生成
tensor1 = torch.linspace(10, 10.8, 5) #steps
## step=(end-start)/(steps-1)
# Example::
#
#         >>> torch.linspace(3, 10, steps=5)
#         tensor([  3.0000,   4.7500,   6.5000,   8.2500,  10.0000])
#         >>> torch.linspace(-10, 10, steps=5)
#         tensor([-10.,  -5.,   0.,   5.,  10.])
#         >>> torch.linspace(start=-10, end=10, steps=5)
#         tensor([-10.,  -5.,   0.,   5.,  10.])
#         >>> torch.linspace(start=-10, end=10, steps=1)
#         tensor([-10.])
#     """
print(tensor1)
```

```python
print(tensor1.dtype)
```

```python
# 3）对数间隔生成
tensor1 = torch.logspace(1, 3, 3)
# step=(end-start)/(steps-1)
# start=1,end=3 step=1
# 10^1,10^2,10^3
# Example::
#
#         >>> torch.logspace(start=-10, end=10, steps=5)
#         tensor([ 1.0000e-10,  1.0000e-05,  1.0000e+00,  1.0000e+05,  1.0000e+10])
#         >>> torch.logspace(start=0.1, end=1.0, steps=5)
#         tensor([  1.2589,   2.1135,   3.5481,   5.9566,  10.0000])
#         >>> torch.logspace(start=0.1, end=1.0, steps=1)
#         tensor([1.2589])
#         >>> torch.logspace(start=2, end=2, steps=1, base=2)
#         tensor([4.0])
print(tensor1)
```

```python
tensor2 = torch.logspace(1, 3, 3, base=2)
print(tensor2)
```

## 3.1.7 随机创建张量

### 模型参数初始化、数据增强、蒙特卡洛模拟

```python
# 1）均匀分布随机张量
# 浮点数
tensor1 = torch.rand(2, 3, dtype=torch.float16)
# Returns a tensor filled with random numbers from a uniform distribution(均匀分布)
#     on the interval :math:`[0, 1)`
print(tensor1)
```

```python
tensor2 = torch.rand_like(tensor1)
print(tensor2)
```

```python
# 整数
tensor3 = torch.randint(10, 100, [5, 7])
#  Returns a tensor filled with random integers generated uniformly
#     between :attr:`low` (inclusive) and :attr:`high` (exclusive).
# 返回一个张量，其中填充了在 low（包含）和 high（不包含）之间均匀生成的随机整数。

print(tensor3)
```

```python
# 2）正态分布随机张量
tensor1 = torch.randn(3, 2, 5)
# Returns a tensor filled with random numbers from a normal distribution
#     with mean `0` and variance `1` (also called the standard normal
#     distribution).
print(tensor1)
```

```python
tensor2 = torch.normal(5, 2, tensor1.shape)

# Returns a tensor of random numbers drawn from separate normal distributions
#     whose mean and standard deviation are given.

print(tensor2)
```

```python
# 3）随机打乱与随机种子
tensor1 = torch.randperm(3)
#Returns a random permutation of integers from ``0`` to ``n - 1``.
print(tensor1)
```
