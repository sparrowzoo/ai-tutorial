# 支持三种索引

```python
import torch
```

```python
tensor1 = torch.randint(1, 10, (3, 5, 6))
print(tensor1)
```

## 3.5.1 简单索引

```python
print(tensor1[0])
print(tensor1[0][0])
print(tensor1[0][0][0])
print(tensor1[0,0,0])
```

```python
print(tensor1[1,2,4])
```

```python
print(tensor1[1][2][4])
```

```python
print(tensor1[1])
```

```python
print(tensor1[1,2])
```

```python
print(tensor1[1, -1])
```

```python
print(tensor1)
print(tensor1[:, 0])
```

## 3.5.2 范围索引

[n:m:s]，n表示起始索引，m表示结束索引，s表示步长。索引范围为[n, m)，不包含m。

```python
print(tensor1[1:])
```

```python
print(tensor1[0])
print(tensor1[0:3:1,0:5:2,1:-1:2])
print(tensor1[-1:])
```

```python
print(tensor1[1:2, :, :])
print(tensor1[1:2, :, :].shape)
#切片不会降维：即使只取 1 个元素，仍保留原维度结构（如 t[1:2, :, :] 形状仍为 (1,5,6)）
print(tensor1[1:2, :, :].squeeze(dim=0).shape)
```

## 3.5.3 列表索引

```python
# 多个列表索引 = 按位置配对！
print(tensor1)
print(tensor1[[2,0],[1,2]])#第三个矩阵取第一行，第0个矩阵取第2行
```

```python
list1 = [0, 1, 1]
list2 = [2, 1, 1]
list3 = [1, 5, 3]
#print(tensor1[list1])
#print(tensor1[list1,list2])
print(tensor1[list1,list2,list3])

```

```python
print(tensor1[[[0], [1]], [1, 2]])
print(tensor1[[[0], [1]], [[1, 2],[1,2]]]) #广播
```

## 3.5.4 布尔索引

```python
# 1. 筛选矩阵（三维张量的元素）
mask = [True, False, True]
print(tensor1[mask])
```

```python
# 筛选满足某种条件的矩阵：中间行、第一个元素大于5的矩阵
mask = tensor1[:, 2, 0] > 5
print(mask)
```

```python
print(tensor1[mask])
```

```python
# 2. 筛选满足条件的元素（具体数值）：大于5的元素
mask = tensor1 > 5
print(mask)
```

```python
print(tensor1[mask])
```

```python
# 3. 筛选满足条件的行：首元素大于5
mask = tensor1[:, :, 0] > 5
print(mask)
```

```python
print(tensor1[mask])
```

```python
# 4. 筛选满足条件的列：首元素大于5
mask = tensor1[:, 0, :] > 5
print(mask)
print(tensor1.mT[mask].mT)
```
