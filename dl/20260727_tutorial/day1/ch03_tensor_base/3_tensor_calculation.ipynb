## 3.3.2 元素级运算（张量与张量）

### 严格一致：当两个张量形状完全相同时，直接按位置对应进行运算。
### 广播机制（Broadcasting）：当形状不同时，PyTorch 会自动将较小的张量扩展为与较大张量相同的形状，再进行逐元素运算。

```python
tensor1 = torch.tensor([[10, 20, 30], [40, 50, 60]])
tensor2 = torch.tensor([[1, 2, 3], [1, 2, 3]])
print(tensor1)
print(tensor2)
```

```python
print(tensor1 + tensor2)
```

```python
print(tensor1 * tensor2)
```

```python
print(tensor1.mul(tensor2))
```

```python
print(tensor1)
```

```python
print(tensor2)
print(tensor1**tensor2)
```

```python
tensor3 = torch.tensor([10, 20, 50])
print(tensor1)
print(tensor3)
```

```python
print(tensor1 + tensor3)
```

## 3.3.3 矩阵相乘

### 3.3.3.1 二维矩阵相乘

```python
tensor1 = torch.tensor(
    [
        [1, 2, 3],
        [4, 5, 6]
    ])
tensor2 = torch.tensor([
    [10,100,1000,10000],
    [20,200,2000,20000],
    [30,300,3000,30000]])
print(tensor1)
print(tensor2)
```

```python
print(tensor1 @ tensor2)

 #    def __rmatmul__(self, other: "Tensor") -> "Tensor":
 #        return torch.matmul(other, self)
```

```python
print(tensor1.matmul(tensor2))
```

```python
print(tensor1.mm(tensor2))
```

```python
tensor3 = torch.tensor([1,2,3])
print(tensor1 @ tensor3)
```

```python
print(tensor1.matmul(tensor3))
print(tensor1.mm(tensor3))# 必须是矩阵
```

### 3.3.3.2 高维张量相乘（批量矩阵乘法）

批量矩阵相乘 就是最右的两维要满足矩阵乘法规则  其他维要满足 广播规则 相当于把内部的矩阵当个元素来处理了 最外层的维度必须得一致

```python
tensor1 = torch.randint(1, 10, (3, 2, 4))
tensor2 = torch.randint(1, 10, (3, 4, 5))
print(tensor1)
print(tensor2)
```

```python
print(tensor1 @ tensor2)
```

```python
print(tensor1.matmul(tensor2))
```

```python
print(tensor1.bmm(tensor2))
```

```python
tensor3 = torch.randn(3, 2, 4, 2)
tensor4 = torch.randn(3, 1, 2, 2)
print(tensor3 @ tensor4)
```
