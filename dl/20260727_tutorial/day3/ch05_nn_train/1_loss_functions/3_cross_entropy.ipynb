```python
import torch
from torch import nn
```

```python
# 定义输出预测值
logits = torch.randn(10, 6)
print(logits)
print(torch.max(logits, dim=-1))
```

```python
# 定义损失函数
loss_fn = nn.CrossEntropyLoss()
```

### 真实值为标签编码

```python
target = torch.randint(0, 6, (10,))
print(target)
```

```python
loss = loss_fn(logits, target)
print(loss)
```

```python
target=torch.argmax(logits, dim=-1)
print(target)
```

```python
loss = loss_fn(logits, target)
print(loss)
print(torch.argmax(logits, dim=-1).eq(target))
```

### 真实值为独热编码

```python
target_onehot = torch.zeros(10, 6)
ten=torch.arange(10)
print(target)
target_onehot[ten, target] = 1
print(target_onehot.argmax(dim=-1))
print(target_onehot)
```

```python
loss = loss_fn(logits, target_onehot)
print(loss)
```
