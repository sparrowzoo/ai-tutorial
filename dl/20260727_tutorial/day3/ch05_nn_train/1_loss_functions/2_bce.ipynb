```python
import torch
from torch import nn
```

```python
# 定义模型输出分数值
logits = torch.randn(5)
print(logits)
```

```python
# 定义真实值（分类标签）
target = torch.randint(0, 2, (5, )).float()
print(target)
```

### BCE

```python
# 定义损失函数
loss_fn = nn.BCELoss()
```

```python
# 将预测数值转换为概率值
proba = torch.sigmoid(logits)
print(proba)
```

```python
# 计算损失
loss = loss_fn(proba, target)
print(loss)
```

### BCEWithLogitsLoss

```python
# 定义损失函数
loss_fn = nn.BCEWithLogitsLoss()

# 计算损失
loss = loss_fn(logits, target)
print(loss)
```
