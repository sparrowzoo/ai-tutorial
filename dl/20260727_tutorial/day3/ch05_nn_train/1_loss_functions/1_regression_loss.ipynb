```python
import torch
from torch import nn
```

```python
# 定义预测评分值（logits）和真实值（target）
logits = torch.randn(5)
target = torch.randn(5)
print(logits)
print(target)
```

## MAE

```python
loss_l1 = nn.L1Loss()
```

```python
loss = loss_l1(logits, target)
print(loss)
```

## MSE

```python
loss_mse = nn.MSELoss(reduction='mean')
```

```python
loss = loss_mse(logits, target)
print(loss)
```

## Smooth L1

```python
loss_smooth = nn.SmoothL1Loss()
```

```python
loss = loss_smooth(logits, target)
print(loss)
```
