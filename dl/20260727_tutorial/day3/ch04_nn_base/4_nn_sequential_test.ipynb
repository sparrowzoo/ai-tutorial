# 定义设备

```python
import torch
from torch import nn
from torchsummary import summary
device = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(device)
print(device=="cpu",device)
```

# 1. 定义输入数据 (10, 3)

```python
X = torch.randn(10, 3).to(device)
print(X)
```

 # 2. 定义模型

```python
model = nn.Sequential(
        nn.Linear(3, 4),
        nn.Tanh(),
        nn.Linear(4, 4),
        nn.ReLU(),
        nn.Linear(4, 2),
        nn.Softmax(dim=-1),
    ).to(device)

print(model)
```

# 3. 前向传播

```python

y = model(X)
print("输出预测值：\n", y)
```

# 4. 查看参数

# 4.1 指定层，逐个查看

```python

print("\n逐个打印参数：")
print("Linear1的参数：")
print("权重：", model.linear1.weight)
print("偏置：", model.linear1.bias)
```

# 4.2 遍历所有层

```python
print("\n遍历打印参数：")
for param in model.parameters():
        print(param)

```

```python
for name, param in model.named_parameters():
        print(name)
        print(param)
        print()
```

# 4.3 查看状态字典

```python
print("状态字典：")
print(model.state_dict())
```

# 5. 查看模型结构和参数数量

```python
print(device=="cpu",device)
print(model)
summary(model, input_size=(3,), batch_size=10,device=device.type)
```

```python
# 6. 保存模型
torch.save(model.state_dict(), 'model.pt')
```

# 7. 加载模型参数

```python

state_dict = torch.load('model.pt')
model.load_state_dict( state_dict )

```

# 8. 使用加载的模型，进行推理预测

```python

x_new = torch.tensor([1, 2, 3]).float().to(device)
y_pred = model(x_new)
print(y_pred)
```
