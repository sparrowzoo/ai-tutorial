# 手写数字识别

````python
import  torch
import  pandas as pd
from   sklearn.model_selection import train_test_split
from   sklearn.preprocessing import MinMaxScaler
````

## 1. 加载数据

````python
samples=pd.read_csv('/Users/zhanglizhi/workspace/sparrow/ai-tutorial/data/train.csv')
print(type(samples))
samples.head()
````

````python
lables=samples.iloc[:,0]
print(type(lables))
print("lables",lables)
# features1=samples.iloc[:,1:-1]
# print(type(features1))
# print("features1 columnes",features1.columns)
features=samples.iloc[:,1:]
print("feature2 column",features.columns)
````

## 2. 划分数据集

````python
train_features,test_features,train_lables,test_lables=train_test_split(features,lables,test_size=0.2,random_state=43)
print("train_features type",type(train_features))
print("train_lables type",type(train_lables))
````

## 3. 特征工程：归一化

````python
scaler=MinMaxScaler()
# print(scaler.data_min_)
# print(scaler.max_)
# scaler.fit(train_features)
# print("scaler.data_min_",scaler.data_min_)
# print("scaler.data_max_",scaler.data_max_)
train_features=scaler.fit_transform(train_features)
test_features=scaler.transform(test_features)
print("train_features type",type(train_features))
print("train_lables type",type(test_features))
````

````python
first_feature=test_features[0]
condition=(first_feature>1)
first_feature[condition]
````

## 4. 转换成Tensor

````python
print("test_lables type",type(test_lables))
test_features_tensor=torch.tensor(test_features).float()
test_lables_tensor=torch.tensor(test_lables.values)
print("train_features type",test_lables_tensor.dtype)
````

````python
first_feature=test_features_tensor[0]
condition=(first_feature<1)&(first_feature>0)
first_feature[condition]
````

## 定义模型
```
1）任务与模型设计
手写数字识别是**10分类任务**（识别0~9共10个数字），我们搭建一个3层全连接神经网络：
输入层：784个神经元（对应28×28=784个像素特征）
隐藏层1：50个神经元，激活函数ReLU
隐藏层2：100个神经元，激活函数ReLU
输出层：10个神经元（对应10个数字类别），训练时无需手动加Softmax（损失函数`CrossEntropyLoss`已内置）
```

````python
digital_model=torch.nn.Sequential(
    torch.nn.Linear(784,50),
    torch.nn.ReLU(),
    torch.nn.Linear(50,100),
    torch.nn.ReLU(),
    torch.nn.Linear(100,10)
)
````

````python
## 加载训练后的模型
dict_state=torch.load("/Users/zhanglizhi/workspace/sparrow/ai-tutorial/data/nn_example.pt",map_location='cpu')
print("训练好的数据字典",dict_state)
digital_model.load_state_dict(dict_state)
````

````python
pred_y=digital_model(test_features_tensor)
print(pred_y.shape)
````
