```python
import torch
from numpy.ma.core import reshape
```

```python
tensor1 = torch.randint(1, 10, (2, 3, 6))
print(tensor1)
```

## 3.6.1 维度交换

```python
# .T：翻转所有维度
print(tensor1.T)
print(tensor1.T.shape)
```

```python
# .mT：矩阵转置
# mT是矩阵转置的简写，仅交换张量的最后两个维度，其余维度保持不变，等价于二维矩阵的转置操作
print(tensor1.mT)
print(tensor1.mT.shape)
```

```python
# .transpose: 交换任意两个维度
print(tensor1.shape)
print(tensor1.transpose(0, 1))
print(tensor1.shape)
```

```python
print(tensor1.transpose(-3, -2))
```

```python
# .permute: 重排列所有维度
print(tensor1) # D,W,H  (1,2,3)=4
print(tensor1[1,2,3])
print(tensor1.permute(2, 0, 1)) #(H,D,W) (3,1,2)
print(tensor1.permute(2, 0, 1)[3,1,2])
```

## 3.6.3 增删维度

```python
tensor2 = torch.tensor([1, 2, 3, 4, 5])
print(tensor2)
print(tensor2.shape)
print(tensor2.dim())
```

```python
tensor3 = tensor2.reshape(1, -1)
print(tensor3)
print(tensor3.shape)
print(tensor3.dim())
```

```python
# 1）增加维度（unsqueeze()方法）
print(tensor2)
tensor3 = tensor2.unsqueeze(dim=0)
print(tensor3)
print(tensor3.shape)
```

```python
print(tensor2)
tensor4 = tensor2.unsqueeze(dim=1)
print(tensor4)
print(tensor4.shape)
```

```python
# 2）删除维度（squeeze()方法）
print(tensor3)
print(tensor3.squeeze(dim=0))
```

```python
print(tensor3)
print(tensor3.squeeze(dim=1))
# squeeze()用于删除所有大小为 1 的维度，也可通过dim指定删除某个位置的维度（仅当该维度大小为 1 时生效）
```

```python
print(tensor4.squeeze(dim=1))
```

```python
tensor2.unsqueeze_(dim=0)
tensor2.unsqueeze_(dim=-1)
print(tensor2)
print(tensor2.shape)
```

```python
print(tensor2.squeeze())
```

## 3.7 张量拼接

```python
# 1）torch.cat()：按指定维度拼接（维度不变）
# 核心规则：除拼接维度外，其他维度必须完全相同，拼接后张量维度数不变，仅拼接维度的大小增加。
tensor1 = torch.randint(1, 10, (3, 2, 5))
tensor2 = torch.randint(1, 10, (3, 1, 5))
print(tensor1)
print(tensor2)
```

```python
tensor3 = torch.cat([tensor1, tensor2], dim=1)
print(tensor3)
print(tensor3.shape)
```

```python
tensor1 = torch.randint(1, 10, (4, 1, 5))
tensor2 = torch.randint(1, 10, (2, 2, 5))
print(tensor1)
print(tensor2)
```

```python
print(torch.cat((tensor1, tensor2),dim=0))
#cat(tensors, dim=0, *, out=None) -> Tensor
```

```python
# 2）torch.stack()：堆叠操作（新增维度）
tensor1 = torch.randint(1, 10, (3, 4, 5))
tensor2 = torch.randint(1, 10, (3, 4, 5))
print(tensor1)
print(tensor2)
```

本质是先给每个张量在指定位置添加一个大小为 1 的维度（unsqueeze），再按该维度拼接（cat），因此堆叠后张量会新增一个维度。
核心要求：所有待堆叠张量的形状必须完全相同。
等价于torch.cat([t1.unsqueeze(0), t2.unsqueeze(0)], dim=0)

```python
tensor3  = torch.stack([tensor1, tensor2], dim=0)
print(tensor3)
print(tensor3.shape)
```
