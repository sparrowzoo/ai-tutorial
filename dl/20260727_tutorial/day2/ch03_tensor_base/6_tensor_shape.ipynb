## 3.7 张量拼接

```python
# 1）torch.cat()：按指定维度拼接（维度不变）
# 核心规则：除拼接维度外，其他维度必须完全相同，拼接后张量维度数不变，仅拼接维度的大小增加。
tensor1 = torch.randint(1, 10, (3, 2, 5))
tensor2 = torch.randint(1, 10, (3, 1, 5))

print(tensor1)
print(tensor2)
print(tensor1.numel()+tensor2.numel())
```

```python
tensor3 = torch.cat([tensor1, tensor2], dim=1)
print(tensor3)
print(tensor3.shape)
print(tensor3.numel())
```

```python
tensor1 = torch.randint(1, 10, (4, 1, 5))
tensor2 = torch.randint(1, 10, (2, 2, 5))
print(tensor1)
print(tensor2)
```

```python
print(torch.cat((tensor1, tensor2),dim=0))
#cat(tensors, dim=0, *, out=None) -> Tensor
```

```python
# 2）torch.stack()：堆叠操作（新增维度）
tensor1 = torch.randint(1, 10, (3, 4, 5))
tensor2 = torch.randint(1, 10, (3, 4, 5))
print(tensor1)
print(tensor2)
```

本质是先给每个张量在指定位置添加一个大小为 1 的维度（unsqueeze），再按该维度拼接（cat），因此堆叠后张量会新增一个维度。
核心要求：所有待堆叠张量的形状必须完全相同。
等价于torch.cat([t1.unsqueeze(0), t2.unsqueeze(0)], dim=0)

```python
tensor3  = torch.stack([tensor1, tensor2], dim=2)
print(tensor3.is_contiguous())
print(tensor3)
print(tensor3.shape)
```

```python
tensor1 = torch.randint(1, 10, (1,1,2,3,1,1,2, 3, 6))
# print(tensor1)
print(tensor1.squeeze().shape)
print(tensor1.unsqueeze(dim=1).shape)
```

## 张量的拼接

```python
import torch
t1 = torch.randint(1, 10, (2, 1, 3))
t2 = torch.randint(1, 10, (2, 3, 3))
print("t1:", t1)
print("t2:", t2)
print(torch.cat([t1, t2], dim=1))
```

```python
t1 = torch.randint(1, 10, (2, 1, 3))
t2 = torch.randint(1, 10, (2, 3, 3))
print("t1:", t1)
print("t2:", t2)
print(torch.cat([t1, t2], dim=1))
```

```python
# 2. 堆叠
t1 = torch.randint(1, 10, (3, 4, 5))
t2 = torch.randint(1, 10, (3, 4, 5))
print("t1:", t1)
print("t2:", t2)
# dim=0 直接把两个张量堆起来
t11 = torch.stack([t1, t2], dim=0)
print("t11",t11)
# dim=1 把两个矩阵堆起来
t12 = torch.stack([t1, t2], dim=1)
print("t12",t12)
# dim=2 把两行堆起来
t13 = torch.stack([t1, t2], dim=2)
print("t13",t13)
# dim=2 把两个标量堆起来
t14 = torch.stack([t1, t2], dim=3)
print("t14",t14)
```

```python
t21=t1.unsqueeze(dim=2)
t22=t2.unsqueeze(dim=2)

print("t21",t21)
print("t22",t22)
print(torch.stack([t1, t2], dim=2))
```
