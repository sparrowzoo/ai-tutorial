```python
import torch
from torch import nn
```

```python
x = torch.tensor([0.7, -1.2, 0, 2.3])
```

## 1. Sigmoid

```python
# 调用函数
y1 = torch.sigmoid(x)
# 调用Tensor的方法
y2 = x.sigmoid()
# 使用nn下的类
sig = nn.Sigmoid()
y3 = sig(x)
print(y1)
print(y2)
print(y3)
```

## Tanh

```python
# 调用函数
y1 = torch.tanh(x)
# 调用Tensor的方法
y2 = x.tanh()
# 使用nn下的类
tanh = nn.Tanh()
y3 = tanh(x)
print(y1)
print(y2)
print(y3)
```

## ReLU

```python
# 调用函数
y1 = torch.relu(x)
# 调用Tensor的方法
y2 = x.relu()
# 使用nn下的类
relu = nn.ReLU()
y3 = relu(x)
print(y1)
print(y2)
print(y3)
```

## Softmax

```python
# 调用函数
y1 = torch.softmax(x, dim=0)
# 调用Tensor的方法
y2 = x.softmax(dim=0)
# 使用nn下的类
softmax = nn.Softmax(dim=0)
y3 = softmax(x)
print(y1)
print(y2)
print(y3)
```

```python
# 测试二维输入
X = torch.tensor([[1.3, -1.5, 0.6], [2.8, 0, -3.2]])
y1 = torch.softmax(X, dim=1)
# 调用Tensor的方法
y2 = X.softmax(dim=0)
# 使用nn下的类
softmax = nn.Softmax(dim=-1)
y3 = softmax(X)
print(y1)
print(y2)
print(y3)
```

## 其它激活函数（ReLU变体）

```python
# LeakyReLU
print(nn.LeakyReLU()(x))
```

```python
# PReLU
print(nn.PReLU()(x))
# RReLU
print(nn.RReLU()(x))
# ELU
print(nn.ELU()(x))
# SiLU
print(nn.SiLU()(x))
# GELU
print(nn.GELU()(x))
# Softplus
print(nn.Softplus()(x))
```
