```python
import torch
import torch.nn as nn
```

## 构造方法

```python
linear = nn.Linear(in_features=2, out_features=3,bias=False)
```

```python
print(linear)
```

```python
# 查看线性层的参数
print("权重：", linear.weight)
print("偏置：", linear.bias)
```

## 前向传播 

```python
# 定义输入数据（2个特征,一个样本）
x = (torch.tensor([2, 2])).float()
print(x)
```

```python
# 调用方法，得到输出y
y = linear.forward(x)
y=linear(x)
print(y)
```

```python
# 定义多条数据（5个样本）
X = torch.randn(5, 2)
print(X)
```

```python
Y = linear.forward(X)
print(Y)
```

PyTorch 中所有 nn.Module 子类（包括 nn.Linear）都重写了 __call__ 方法，__call__ 会自动调用 forward 方法，因此可以简化写法：
等价于 linear.forward(x)，PyTorch 标准写法
y = linear(x)

```python
# 隐式调用
y = linear(x)
print(y)
Y = linear(X)
print(Y)
```
