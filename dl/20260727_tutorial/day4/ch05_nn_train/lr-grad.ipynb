```python
import  torch
import torch.nn as nn
import torch.optim as optim

weight=torch.tensor([[1,2],[2,3],[3,4.0]])
print(weight.shape)
class LR(nn.Module):
    def __init__(self):
        super().__init__()
        self.linear = nn.Linear(2, 3)
        self.linear.weight.data=weight
        self.linear.bias.data=torch.tensor([0,1,0],dtype=torch.float)
    def forward(self, x):
        return self.linear(x)

lr = LR()
x=torch.tensor([[1,2],[2,3]],dtype=torch.float)
preds=lr.forward(x)
print(preds.shape)
```

```python
loss_fu=nn.CrossEntropyLoss(reduction='mean')
optim=optim.SGD(lr.parameters(),lr=0.1)
```

```python
labels = torch.tensor([[0, 0, 1],[0,1,0]], dtype=torch.float)
loss=loss_fu(preds,labels)
print(loss)
loss.backward()
```

```python
print(lr.linear.weight.grad.T)
print(lr.linear.bias.grad)
```

```python
probs=torch.softmax(preds,dim=-1)
print(probs)
```

```python
loss = (- labels.float()*(torch.log(probs))).sum()/2
print(loss)
```

```python
print(x.shape)
E = (probs - labels)
print(x.T@E/2)
```
