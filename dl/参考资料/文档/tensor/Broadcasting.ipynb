# 广播语义
许多PyTorch操作支持NumPy的广播语义。
详情请见 https://numpy.org/doc/stable/user/basics.broadcasting.html。
简而言之，如果一个PyTorch操作支持广播，那么它的Tensor参数可以自动扩展为相同的大小（无需复制数据）。


## 通用广播规则
```
General broadcasting rules
When operating on two arrays, NumPy compares their shapes element-wise.
 It starts with the trailing (i.e. rightmost)
dimension and works its way left. Two dimensions are compatible when

they are equal, or
one of them is 1.
```
---
📌 总结
“从右向左，缺则补 1，相同或 1 则过，否则报错；结果取大，两边都扩”

# 示例

## AB双广播

````python
import torch
A = torch.tensor([[[1, 2, 3, 4]],  # shape (3, 1, 4) 最大取 3,2,4
                  [[21, 22, 33, 44]],
                  [[31, 32, 33, 34]]])
print("A shape=",A.shape)
B = torch.tensor([[10, 20, 30, 40],    # shape (2, 4) =>(1,2,4) 取最大为 3,2,4
                  [100, 200, 300, 400]])
print("B shape=",B.shape)
C = A + B
print(C)
print("C shape=",C.shape)
````

````python
A_expanded = [
    # 第 0 层（原始 A[0]）
    [[1, 2, 3, 4],      # 原始行
     [1, 2, 3, 4]],     # 复制出来的行（和上面一模一样）

    # 第 1 层（原始 A[1]）
    [[21, 22, 33, 44],  # 原始行
     [21, 22, 33, 44]], # 复制出来的行

    # 第 2 层（原始 A[2]）
    [[31, 32, 33, 34],  # 原始行
     [31, 32, 33, 34]]  # 复制出来的行
]
A_expanded = torch.tensor(A_expanded)
B_expanded = [
    # 第 0 层（原始 B 的第 0 行和第 1 行）
    [[10, 20, 30, 40],   # 原始 B[0]
     [100, 200, 300, 400]], # 原始 B[1]

    # 第 1 层（完全复制第 0 层）
    [[10, 20, 30, 40],
     [100, 200, 300, 400]],

    # 第 2 层（完全复制第 0 层）
    [[10, 20, 30, 40],
     [100, 200, 300, 400]]
]
B_expanded = torch.tensor(B_expanded)

print("A_expanded shape=",A_expanded.shape)
print("B_expanded shape=",B_expanded.shape)

C_EXPANDED = A_expanded + B_expanded
print("C_expanded shape=",C_EXPANDED)

print(C==C_EXPANDED)
````
