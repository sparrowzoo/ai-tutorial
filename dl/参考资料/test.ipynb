```python
import torch
xy=torch.randint(10,20,(100,1))
```

```python
h1=torch.tensor(0)
h2=torch.tensor(0)
h1_history=[]
h2_history=[]
for x in xy:
    h1=h1+x
    h1_history.append(torch.sqrt(h1).item())
    h2=h2+torch.sqrt(x)
    h2_history.append(h2.item())
import matplotlib.pyplot as plot
plot.plot(h1_history,label='h1')
plot.plot(h2_history,label='h2')
plot.legend()
plot.show()
```
