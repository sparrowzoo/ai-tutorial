# 画线

```python
from matplotlib import pyplot as plt
from sklearn.externals.array_api_compat import torch
array=torch.randint(1, 100, (100,)).numpy()
array2=torch.randint(1, 100, (100,)).numpy()
plt.plot(array, 'r', label='train accuracy')
plt.plot(array2, 'b', label='test accuracy')
plt.legend(loc='upper right')
plt.xlabel('epoch')
plt.ylabel('accuracy')
plt.show()
```

# 画多图 画点

```python

fig, ax = plt.subplots(1, 2, figsize=(12, 4))
indexs= torch.arange(len(array.data)).numpy()
# 1. 训练误差随轮次下降曲线
## 按数据索引位置自动添充x轴
ax[0].plot(array, 'r', label='train accuracy')
## x轴代表位置
ax[0].set_xlabel('epoch')
## y轴代表损失
ax[0].set_ylabel('loss')
# 2. 散点图
ax[1].scatter(indexs,array2)
ax[1].plot(array2, 'b', label='test accuracy')
ax[1].plot(indexs, 'r', label='test index')
#拟合直线
plt.show()
```

# 画等高线
本质是 X、Y 分别广播到 [len(X), len(Y)] 矩阵，然后按元素矩阵相乘，得到函数值矩阵 Z。
这就是 meshgrid + 函数计算的完整数学本质。

```python
import numpy as np
import matplotlib.pyplot as plt
def f(w):
    #  是一个包含两个元素的张量 [w0, w1]
    return 0.05 * w[0]**2 + w[1] ** 2
w0=np.linspace(-7, 7, 15)
w1=np.linspace(-2, 2, 5)
W0,W1=np.meshgrid(w0, w1)
print(W0.shape)
print(W1.shape)
loss = f([W0, W1])
print("W0",W0)
print("W1",W1)
print(loss[0])
sgd_w0_list=[-1,0,-2]
sgd_w1_list=[-2,-1,1]
plt.xlabel('w0')
plt.ylabel('w1')
plt.plot(sgd_w0_list, sgd_w1_list, color='r')
contour=plt.contour(W0, W1, loss, levels=30)
# plt.clabel(contour, inline=True, fontsize=8)
# plt.colorbar()  # 显示颜色条
```
